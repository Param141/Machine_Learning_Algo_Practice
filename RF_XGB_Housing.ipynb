### Handling Missing Values

Based on our EDA, several columns have a very high percentage of missing values. We will drop these columns first. Then, we will address the remaining missing values with appropriate imputation strategies.

```python
# Columns to drop due to high percentage of missing values (e.g., > 80%)
columns_to_drop = missing_percentage[missing_percentage > 80].index.tolist()
print(f"Dropping columns: {columns_to_drop}")

df_train_processed = df_train.drop(columns=columns_to_drop)

print(f"Shape of dataframe after dropping columns: {df_train_processed.shape}")
```

Next, let's re-evaluate the missing values in the remaining columns to determine our next imputation strategy. We will also address the skewness of the `SalePrice` target variable.

```python
# Re-check missing values after dropping columns
missing_data_after_drop = df_train_processed.isnull().sum()
missing_data_after_drop = missing_data_after_drop[missing_data_after_drop > 0].sort_values(ascending=False)

print('Missing values after dropping high-percentage columns:')
display(missing_data_after_drop)

# Log transform SalePrice
df_train_processed['SalePrice'] = np.log1p(df_train_processed['SalePrice'])

plt.figure(figsize=(10, 6))
sns.histplot(df_train_processed['SalePrice'], kde=True, bins=50)
plt.title('Distribution of Log-transformed SalePrice')
plt.xlabel('Log(SalePrice)')
plt.ylabel('Frequency')
plt.show()
```

### Imputing Remaining Missing Values

We will now impute the remaining missing values. Numerical features will be filled with their respective medians, and categorical features with their modes or a specific 'None' category where applicable (e.g., for features describing absence).

```python
# Identify numerical and categorical columns with missing values
numerical_cols_with_na = df_train_processed.select_dtypes(include=['int64', 'float64']).columns[df_train_processed.select_dtypes(include=['int64', 'float64']).isnull().any()].tolist()
categorical_cols_with_na = df_train_processed.select_dtypes(include='object').columns[df_train_processed.select_dtypes(include='object').isnull().any()].tolist()

print(f"Numerical columns with NA: {numerical_cols_with_na}")
print(f"Categorical columns with NA: {categorical_cols_with_na}")

# Impute numerical columns with their median
for col in numerical_cols_with_na:
    # Special handling for GarageYrBlt if missing implies no garage
    if col == 'GarageYrBlt':
        df_train_processed[col] = df_train_processed[col].fillna(0) # Fill with 0 for no garage
    else:
        median_val = df_train_processed[col].median()
        df_train_processed[col] = df_train_processed[col].fillna(median_val)
        print(f"Imputed numerical column '{col}' with median: {median_val}")

# Impute categorical columns with 'None' where absence is indicated or with the mode
for col in categorical_cols_with_na:
    if col in ['MasVnrType', 'FireplaceQu', 'GarageType', 'GarageFinish', 'GarageQual', 'GarageCond',
               'BsmtExposure', 'BsmtFinType2', 'BsmtCond', 'BsmtQual', 'BsmtFinType1']:
        df_train_processed[col] = df_train_processed[col].fillna('None') # Fill with 'None' for absence
        print(f"Imputed categorical column '{col}' with 'None'")
    else:
        mode_val = df_train_processed[col].mode()[0]
        df_train_processed[col] = df_train_processed[col].fillna(mode_val) # Fill with mode for other categoricals
        print(f"Imputed categorical column '{col}' with mode: {mode_val}")

# Verify no more missing values
print('\nMissing values after imputation:')
display(df_train_processed.isnull().sum().sum())
```

### Feature Engineering and Encoding Categorical Variables

Now that missing values are handled, we need to convert our categorical features into a numerical format suitable for machine learning models. We will use one-hot encoding for this.

```python
# Separate features into numerical and categorical for encoding
numerical_features = df_train_processed.select_dtypes(include=['int64', 'float64']).columns.tolist()
categorical_features = df_train_processed.select_dtypes(include='object').columns.tolist()

print(f"Numerical Features ({len(numerical_features)}):\n{numerical_features}")
print(f"Categorical Features ({len(categorical_features)}):\n{categorical_features}")

# One-Hot Encode categorical features
df_train_encoded = pd.get_dummies(df_train_processed, columns=categorical_features, drop_first=True)

print('\nShape of dataframe after one-hot encoding:')
print(df_train_encoded.shape)
display(df_train_encoded.head())
```

### Model Training: RandomForestRegressor

Now that our data is prepared, let's train a `RandomForestRegressor` model to predict house sale prices. We will use the `X_train` and `y_train` datasets for training.

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score

# Initialize the RandomForestRegressor model
# Using a small number of estimators for initial run, can be tuned later
rf_model = RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1)

# Train the model
rf_model.fit(X_train, y_train)

print("RandomForestRegressor model trained successfully!")
```

### Model Evaluation

After training, it's crucial to evaluate our model's performance on both the training and testing datasets. We will use metrics like Mean Squared Error (MSE) and R-squared ($R^2$) to assess how well the model predicts `SalePrice`.

```python
# Make predictions on the training set
y_train_pred_rf = rf_model.predict(X_train)

# Evaluate training set performance
mse_train_rf = mean_squared_error(y_train, y_train_pred_rf)
r2_train_rf = r2_score(y_train, y_train_pred_rf)

print(f"Training Mean Squared Error (RandomForest): {mse_train_rf:.4f}")
print(f"Training R-squared (RandomForest): {r2_train_rf:.4f}")

# Make predictions on the test set
y_test_pred_rf = rf_model.predict(X_test)

# Evaluate test set performance
mse_test_rf = mean_squared_error(y_test, y_test_pred_rf)
r2_test_rf = r2_score(y_test, y_test_pred_rf)

print(f"Test Mean Squared Error (RandomForest): {mse_test_rf:.4f}")
print(f"Test R-squared (RandomForest): {r2_test_rf:.4f}")

# Plotting actual vs predicted values for the test set
plt.figure(figsize=(10, 6))
sns.scatterplot(x=y_test, y=y_test_pred_rf, alpha=0.6)
plt.plot([min(y_test), max(y_test)], [min(y_test), max(y_test)], color='red', linestyle='--')
plt.title('RandomForestRegressor: Actual vs. Predicted SalePrice (Test Set)')
plt.xlabel('Actual Log SalePrice')
plt.ylabel('Predicted Log SalePrice')
plt.grid(True)
plt.show()
```

### Hyperparameter Tuning with RandomizedSearchCV

To improve our model's performance and generalize better to unseen data, we'll perform hyperparameter tuning using `RandomizedSearchCV`. This will help us find an optimal combination of hyperparameters for the `RandomForestRegressor`.

```python
from sklearn.model_selection import RandomizedSearchCV

# Define the parameter grid to sample from
param_dist = {
    'n_estimators': [100, 200, 300, 400, 500], # Number of trees in the forest
    'max_features': ['sqrt', 'log2', 0.6, 0.8], # Number of features to consider at every split
    'max_depth': [10, 20, 30, 40, 50, None], # Maximum number of levels in tree
    'min_samples_split': [2, 5, 10], # Minimum number of samples required to split a node
    'min_samples_leaf': [1, 2, 4], # Minimum number of samples required at each leaf node
    'bootstrap': [True, False] # Method for sampling data points (with or without replacement)
}

# Initialize a new RandomForestRegressor
rf = RandomForestRegressor(random_state=42, n_jobs=-1)

# Initialize RandomizedSearchCV
# n_iter: Number of parameter settings that are sampled. Trades off execution time with quality of solution.
# cv: Number of folds for cross-validation.
rf_random = RandomizedSearchCV(estimator=rf, param_distributions=param_dist,
                               n_iter=50, cv=3, verbose=2, random_state=42, n_jobs=-1, scoring='neg_mean_squared_error')

# Fit the random search model
rf_random.fit(X_train, y_train)

print("RandomizedSearchCV completed. Best parameters found:")
display(rf_random.best_params_)

# Get the best model
best_rf_model = rf_random.best_estimator_

print('\nRe-evaluating the model with best parameters:')
```

```python
# Make predictions on the training set with the best model
y_train_pred_tuned = best_rf_model.predict(X_train)

# Evaluate training set performance
mse_train_tuned = mean_squared_error(y_train, y_train_pred_tuned)
r2_train_tuned = r2_score(y_train, y_train_pred_tuned)

print(f"Tuned Training Mean Squared Error (RandomForest): {mse_train_tuned:.4f}")
print(f"Tuned Training R-squared (RandomForest): {r2_train_tuned:.4f}")

# Make predictions on the test set with the best model
y_test_pred_tuned = best_rf_model.predict(X_test)

# Evaluate test set performance
mse_test_tuned = mean_squared_error(y_test, y_test_pred_tuned)
r2_test_tuned = r2_score(y_test, y_test_pred_tuned)

print(f"Tuned Test Mean Squared Error (RandomForest): {mse_test_tuned:.4f}")
print(f"Tuned Test R-squared (RandomForest): {r2_test_tuned:.4f}")

# Plotting actual vs predicted values for the test set with the tuned model
plt.figure(figsize=(10, 6))
sns.scatterplot(x=y_test, y=y_test_pred_tuned, alpha=0.6)
plt.plot([min(y_test), max(y_test)], [min(y_test), max(y_test)], color='red', linestyle='--')
plt.title('Tuned RandomForestRegressor: Actual vs. Predicted SalePrice (Test Set)')
plt.xlabel('Actual Log SalePrice')
plt.ylabel('Predicted Log SalePrice')
plt.grid(True)
plt.show()
```
